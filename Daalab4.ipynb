```python
#program 1 graph+code

import time
import matplotlib.pyplot as plt


def max_product(arr):

    max_product = arr[0]
    max_current = arr[0]
    min_current = arr[0]

    for i in range(1, len(arr)):

        if arr[i] < 0:
            max_current, min_current = min_current, max_current

        max_current = max(arr[i], max_current * arr[i])
        min_current = min(arr[i], min_current * arr[i])

        if max_current > max_product:
            max_product = max_current

    return max_product


# Taking input from user
n = int(input("Enter number of elements: "))

arr = []

for i in range(n):
    arr.append(int(input("Enter element: ")))

print("Maximum Product:", max_product(arr))


# Graph
sizes = [100, 500, 1000, 2000, 5000]
times = []

for n in sizes:

    test_arr = []

    for i in range(n):
        test_arr.append((i % 10) + 1)

    start = time.time()

    max_product(test_arr)

    end = time.time()

    times.append(end - start)


plt.plot(sizes, times, marker='o')

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Maximum Product Subarray")

plt.grid()
plt.show()
```

```python
#program2
def partition(arr, low, high):
    pivot = arr[high]
    i = low

    for j in range(low, high):
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]

    return i


def quickselect(arr, low, high, k):

    if low <= high:

        pos = partition(arr, low, high)

        if pos == k:
            return arr[pos]

        elif pos > k:
            return quickselect(arr, low, pos - 1, k)

        else:
            return quickselect(arr, pos + 1, high, k)


n = int(input("Enter number of elements: "))

arr = []

for i in range(n):
    arr.append(int(input("Enter element: ")))

k = int(input("Enter k: "))

index = n - k

answer = quickselect(arr, 0, n - 1, index)

print("Kth Largest Element:", answer)
```

```python
import time
import matplotlib.pyplot as plt
import random

sizes = [100, 500, 1000, 2000, 5000]
times = []

for n in sizes:

    arr = []

    for i in range(n):
        arr.append(random.randint(1, 10000))

    k = n // 2
    index = n - k

    start = time.time()

    quickselect(arr, 0, n - 1, index)

    end = time.time()

    times.append(end - start)


print("Input Sizes:", sizes)
print("Execution Times:", times)

plt.plot(sizes, times, marker='o')

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Quick Select Performance")

plt.grid()
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

sizes = [100, 500, 1000, 2000, 5000]
times = []

for n in sizes:

    arr = []

    for i in range(n):
        arr.append(i + 1)

    start = time.time()

    i = 0

    while i < n:

        position = arr[i] - 1

        if arr[i] > 0 and arr[i] <= n and arr[i] != arr[position]:

            arr[i], arr[position] = arr[position], arr[i]

        else:
            i += 1


    for i in range(n):

        if arr[i] != i + 1:
            break

    end = time.time()

    times.append(end - start)


plt.plot(sizes, times, marker='o')

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Smallest Missing Positive")

plt.grid()
plt.show()
```
