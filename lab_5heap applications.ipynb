```python
#program1 heap sort
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)


def heap_sort(arr):
    n = len(arr)

    # Build max heap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Extract elements one by one
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        heapify(arr, i, 0)


arr = [12, 11, 13, 5, 6]

heap_sort(arr)

print("Sorted array:", arr)
```

```python
#program2rearrange string
import heapq
from collections import Counter

def rearrange_string(s):
    count = Counter(s)

    # Check if rearrangement is possible
    if max(count.values()) > (len(s) + 1) // 2:
        return "Not possible"

    # Max heap using negative frequency
    heap = [(-freq, char) for char, freq in count.items()]
    heapq.heapify(heap)

    result = []
    prev_freq = 0
    prev_char = ""

    while heap:
        freq, char = heapq.heappop(heap)

        result.append(char)

        # Put previous character back
        if prev_freq < 0:
            heapq.heappush(heap, (prev_freq, prev_char))

        freq += 1
        prev_freq = freq
        prev_char = char

    return "".join(result)


s = "aaabbc"

print("Original string:", s)
print("Rearranged string:", rearrange_string(s))
```

```python
#program3
import heapq

arrays = [[1, 4, 7], [2, 5, 8], [3, 6, 9]]

heap = []
result = []

# Add first element of each array
for i in range(len(arrays)):
    heapq.heappush(heap, (arrays[i][0], i, 0))

# Merge arrays
while heap:
    value, i, j = heapq.heappop(heap)
    result.append(value)

    if j + 1 < len(arrays[i]):
        heapq.heappush(heap, (arrays[i][j + 1], i, j + 1))

print("Input:", arrays)
print("Merged Array:", result)
```

```python
#program 4
def median_of_medians(arr, k):

    # Divide into groups of 5
    groups = [arr[i:i+5] for i in range(0, len(arr), 5)]

    # Find median of each group
    medians = []
    for group in groups:
        group.sort()
        medians.append(group[len(group)//2])

    # Find median of medians
    medians.sort()
    pivot = medians[len(medians)//2]

    # Divide around pivot
    smaller = [x for x in arr if x < pivot]
    larger = [x for x in arr if x > pivot]

    if k == len(smaller) + 1:
        return pivot
    elif k <= len(smaller):
        return median_of_medians(smaller, k)
    else:
        return median_of_medians(larger, k - len(smaller) - 1)


arr = [7, 10, 4, 3, 20, 15]
k = 3

print("Kth smallest element:", median_of_medians(arr, k))
```
