```python
import time
import matplotlib.pyplot as plt

def matrix_multiply(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]

    return C


sizes = [10, 20, 30, 40, 50]
times = []

for n in sizes:
    A = [[1] * n for _ in range(n)]
    B = [[1] * n for _ in range(n)]

    start = time.perf_counter()
    matrix_multiply(A, B)
    end = time.perf_counter()

    times.append(end - start)

print("Sizes:", sizes)
print("Times:", times)

plt.plot(sizes, times, marker='o')
plt.xlabel("Matrix Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Matrix Multiplication - O(n³)")
plt.grid()
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

def allocate_pages(books, students):

    low = max(books)
    high = sum(books)

    while low <= high:

        mid = (low + high) // 2

        count = 1
        pages = 0

        for book in books:

            if pages + book <= mid:
                pages += book
            else:
                count += 1
                pages = book

        if count <= students:
            high = mid - 1
        else:
            low = mid + 1

    return low


sizes = [10, 20, 40, 80, 160, 320]
times = []

for n in sizes:

    books = [100] * n
    students = n // 2#

    start = time.perf_counter()
    allocate_pages(books, students)
    end = time.perf_counter()

    times.append(end - start)

print("Number of Books:", sizes)
print("Times:", times)

plt.plot(sizes, times, marker='o')
plt.xlabel("Number of Books (n)")
plt.ylabel("Time (seconds)")
plt.title("Allocate Minimum Pages - O(n log S)")
plt.grid()
plt.show()
```

```python
#code3
A = [1, 3]
B = [2]

# A should be the smaller array
if len(A) > len(B):
    A, B = B, A

m = len(A)
n = len(B)

low = 0
high = m

while low <= high:

    mid = (low + high) // 2

    # Partition of B
    j = (m + n + 1) // 2 - mid

    if mid == 0:
        leftA = float('-inf')
    else:
        leftA = A[mid - 1]

    if mid == m:
        rightA = float('inf')
    else:
        rightA = A[mid]

    if j == 0:
        leftB = float('-inf')
    else:
        leftB = B[j - 1]

    if j == n:
        rightB = float('inf')
    else:
        rightB = B[j]

    # Correct partition
    if leftA <= rightB and leftB <= rightA:

        if (m + n) % 2 == 0:
            median = (max(leftA, leftB) + min(rightA, rightB)) / 2
        else:
            median = max(leftA, leftB)

        break

    elif leftA > rightB:
        high = mid - 1

    else:
        low = mid + 1

print("Median =", median)
```
