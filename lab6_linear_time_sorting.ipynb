```python

import matplotlib.pyplot as plt
import time
import random

def counting_sort(arr):
    count = [0] * (max(arr) + 1)

    for x in arr:
        count[x] += 1

    result = []
    for i in range(len(count)):
        result.extend([i] * count[i])

    return result

sizes = [1000, 2000, 3000,4000]
times = []

for n in sizes:
    arr = [random.randint(0, 100) for _ in range(n)]

    start = time.perf_counter()
    counting_sort(arr)
    end = time.perf_counter()

    times.append(end - start)

print("Input Sizes:", sizes)
print("Execution Times:", times)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Counting Sort Time Complexity")
plt.grid(True)
plt.show()
```

```python
import time
import random
import matplotlib.pyplot as plt

def bucket_sort(arr):
    n = len(arr)
    buckets = [[] for i in range(n)]

    for x in arr:
        index = int(x * n)
        buckets[index].append(x)

    for bucket in buckets:
        bucket.sort()

    result = []

    for bucket in buckets:
        result.extend(bucket)

    return result


sizes = [100, 200, 300, 400, 500]
times = []

for n in sizes:
    arr = [random.random() for i in range(n)]

    start = time.perf_counter()
    bucket_sort(arr)
    end = time.perf_counter()

    times.append(end - start)

print("Input Sizes:", sizes)
print("Execution Times:", times)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size")
plt.ylabel("Execution Time")
plt.title("Bucket Sort Time Complexity")
plt.grid()
plt.show()
```

```python
import time
import random
import matplotlib.pyplot as plt

def at_most(arr, k):
    count = {}
    left = 0
    ans = 0

    for right in range(len(arr)):
        x = arr[right]
        count[x] = count.get(x, 0) + 1

        while len(count) > k:
            count[arr[left]] -= 1

            if count[arr[left]] == 0:
                del count[arr[left]]

            left += 1

        ans += right - left + 1

    return ans


def exactly_k(arr, k):
    return at_most(arr, k) - at_most(arr, k - 1)


sizes = [100, 200, 300, 400, 500]
times = []

for n in sizes:
    arr = [random.randint(1, 20) for i in range(n)]

    start = time.perf_counter()
    exactly_k(arr, 5)
    end = time.perf_counter()

    times.append(end - start)

print("Input Sizes:", sizes)
print("Execution Times:", times)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size")
plt.ylabel("Execution Time")
plt.title("Exactly K Distinct Time Complexity")
plt.grid()
plt.show()
```

```python
import heapq
import time
import matplotlib.pyplot as plt

def minimize_sum(arr, k):
    heap = [-x for x in arr]
    heapq.heapify(heap)

    for i in range(k):
        x = -heapq.heappop(heap)
        x = (x + 1) // 2
        heapq.heappush(heap, -x)

    return sum(-x for x in heap)


arr = [100000, 80000, 60000, 40000, 20000]

sizes = [100, 200, 300, 400, 500]
times = []

for k in sizes:
    start = time.perf_counter()

    minimize_sum(arr, k)

    end = time.perf_counter()

    times.append(end - start)

print("K Values:", sizes)
print("Execution Times:", times)

plt.plot(sizes, times, marker='o')
plt.xlabel("Number of Operations (K)")
plt.ylabel("Execution Time")
plt.title("Minimize Array Sum - Time Complexity")
plt.grid()
plt.show()
```
